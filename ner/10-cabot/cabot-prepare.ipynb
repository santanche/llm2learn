```python
import pandas as pd
import cabot_diagnosis
import cabot_ner
```

```python
INPUT_CABOT = "../../data/case-teaching-cabot/case-teaching-cabot.csv"
OUTPUT_DIAGNOSIS = "../../data/case-teaching-cabot/case-teaching-cabot-with-diagnosis.csv"
OUTPUT_NER = "../../data/case-teaching-cabot/case-teaching-cabot-with-ner.csv"
```

```python
# -----------------------------
# Load CSV
# -----------------------------
df = pd.read_csv(INPUT_CABOT)

if not {"number", "case"}.issubset(df.columns):
    raise ValueError("CSV must contain 'number' and 'case' columns")

# -----------------------------
# Apply NER
# -----------------------------
df["case_ner"] = df["case"].apply(cabot_ner.extract_ner_entities)
df["case_ner_summary"] = df["case_ner"].apply(cabot_ner.build_ner_summary)
df["diagnosis"] = df["case"].apply(cabot_diagnosis.extract_diagnosis)
df["diagnosis_ner_summary"] = df["diagnosis"].apply(cabot_ner.extract_ner_entities).apply(cabot_ner.build_ner_summary)

df_diagnosis = pd.DataFrame({
    "number": df["number"],
    "diagnosis": df["diagnosis"],
    "diagnosis_ner_summary": df["diagnosis_ner_summary"]
})

# -----------------------------
# Save result
# -----------------------------
df.to_csv(
    OUTPUT_NER,
    index=False,
    encoding="utf-8",
    quoting=1  # csv.QUOTE_ALL (safe for JSON-like columns)
)
print(f"NER completed. Output saved to {OUTPUT_NER}")

df_diagnosis.to_csv(
    OUTPUT_DIAGNOSIS,
    index=False,
    encoding="utf-8",
    quoting=1  # csv.QUOTE_ALL (safe for JSON-like columns)
)
print(f"Saved {len(df_diagnosis)} rows to {OUTPUT_DIAGNOSIS}")
```
